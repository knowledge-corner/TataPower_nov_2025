### Decision Making - if-else

###### Ex. WAP to check if a number is even or odd 

```python
num = int(input("Enter a number - "))
if num % 2 == 0 :
    print("Even Number")
else:
    print("Odd Number")
```

###### Ex. Write a Python program to find those numbers which are divisible by 7 and 5

```python
num = int(input("Enter a number - "))
if (num % 5 == 0) and (num % 7 == 0) :
    print("DIvisible by 5 & 7")
else:
    print("Not divisible")
```

```python
num = int(input("Enter a number - "))
if (num % 5 == 0) and (num % 7 == 0) :
    print("Divisible by 5 & 7")
elif num % 5 == 0 :
    print("Divisible by only 5")
elif num % 7 == 0 :
    print("Divisible by only 7")
else:
    print("Not divisible by either")
```

###### Ex. WAP to check a user has entered a single character, then check if it is a vowel

```python
ch = input("Enter a single character - ").lower()
if ch in "aeiouAEIOU" :
    print("Vowel")
else:
    print("Consonant")
```

```python
ch = input("Enter a single character - ").lower()
if len(ch) == 1 and ch.isalpha():
    if ch in "aeiou" :
        print("Vowel")
    else:
        print("Consonant")
else:
    print("Invalid input")
```

### For - loop

###### Ex. WAP to print square os numbers in the list - 

```python
numbers = [1, 4, 6, 2, 8]
for var in numbers :
    print(var, " - ", var**2)
```

###### Ex. WAP to print square of only the even numbers from the list

```python
numbers = [1, 4, 6, 2, 8, 3, 5]
for var in numbers :
    if var % 2 == 0 :
        print(var, " - ", var**2)
```
